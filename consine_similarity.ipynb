```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity
text = ["London Paris London", "Paris Paris London"]
#sci-kit learn has an impmentation to figure out the count the number of words and their frequencies as well
cv = CountVectorizer()

count_matrix = cv.fit_transform(text)
# fit_transform does two things:
# 1. fit(): Extracts the unique vocabulary from the text (e.g., "london", "paris").
# 2. transform(): Creates a matrix where each row corresponds to a string in 'text',
#    and each column represents the count of a vocabulary word in that string.
#    e.g., [2, 1] means "london" appears 2 times, "paris" appears 1 time.

print(count_matrix.toarray())

# count_matrix is a "sparse matrix" (it saves memory by not storing zeros).
# So we use .toarray() to convert it into a standard 2D array so it's easier to read and print.

similarity_scores = cosine_similarity(count_matrix)
print(similarity_scores)

```

#### Euclidean Distance vs. Cosine Similarity

1. Euclidean Distance (Straight-line distance)
   - Best used when the sheer *magnitude* or specific values are important (e.g., comparing physical height/weight).
   - Because it calculates absolute distance, it is highly sensitive to magnitude. If two things have a similar pattern but vastly different sizes, Euclidean distance will flag them as different.

2. Cosine Similarity (Angular distance)
   - Best used when the *pattern* or *orientation* matters more than the magnitude (e.g., comparing text documents or movie genres).
   - Because it measures the angle between vectors, it is invariant to magnitude. If one movie review is a 10-page essay and another is a single paragraph, but they use similar vocabulary proportionally, Cosine Similarity will correctly recognize them as highly similar.
